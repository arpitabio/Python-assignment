```python
#Mean
gene_exp_mean = df["Gene_Expression"].mean()
gene_exp_A_mean = df["Gene_A_Expression"].mean()
gene_exp_B_mean = df["Gene_B_Expression"].mean()
print("Mean:")
print("Gene_Expression:", gene_exp_mean)
print("Gene_A_Expression:", gene_exp_A_mean)
print("Gene_B_Expression:", gene_exp_B_mean)
```

```python
#Median
gene_exp_median = df["Gene_Expression"].median()
gene_exp_A_median = df["Gene_A_Expression"].median()
gene_exp_B_median = df["Gene_B_Expression"].median()
print("Median:")
print("Gene_Expression:", gene_exp_median)
print("Gene_A_Expression:", gene_exp_A_median)
print("Gene_B_Expression:", gene_exp_B_median)
```

```python
#Standard Deviation
gene_exp_std = df["Gene_Expression"].std()
gene_exp_A_std = df["Gene_A_Expression"].std()
gene_exp_B_std = df["Gene_B_Expression"].std()
print("Standard Deviation:")
print("Gene_Expression:", gene_exp_std)
print("Gene_A_Expression:", gene_exp_A_std)
print("Gene_B_Expression:", gene_exp_B_std)
```

```python
#Creating the histogram plot
import matplotlib.pyplot as plt

# Create a figure and axis
plt.figure(figsize=(10, 6))

# Plot histograms for Gene_Expression, Gene_A_Expression, and Gene_B_Expression
plt.hist(df["Gene_Expression"], bins=10, color='blue', alpha=0.5,edgecolor='black', label='Gene_Expression')
plt.hist(df["Gene_A_Expression"], bins=10, color='green', alpha=0.5,edgecolor='black', label='Gene_A_Expression')
plt.hist(df["Gene_B_Expression"], bins=10, color='red', alpha=0.5,edgecolor='black', label='Gene_B_Expression')

# Add a legend
plt.legend()

# Add titles and labels
plt.title('Distribution of Gene Expression Levels')
plt.xlabel('Expression Level')
plt.ylabel('Frequency')

# Show the plot
plt.show()
```

```python
#Creating the distribution of gene expression with histogram and density plots
import numpy as np
import matplotlib.pyplot as plt

# List of gene expression columns
columns = ['Gene_Expression', 'Gene_A_Expression', 'Gene_B_Expression']

# Set the number of subplots
num_columns = len(columns)

# Set the figure size for all plots
plt.figure(figsize=(18, 6))

# Loop through each column to create histograms and density plots
for i, col in enumerate(columns):
    plt.subplot(1, num_columns, i + 1)  # Create a subplot for each column

    # Create a histogram
    plt.hist(df[col], bins=30, density=True, color='blue', alpha=0.5, label='Histogram')

    # Calculate density using NumPy
    data = df[col]
    x = np.linspace(data.min(), data.max(), 1000)
    density = np.histogram(data, bins=30, density=True)[0]
    
    # Calculate bin centers for density plot
    bin_centers = 0.5 * (np.histogram(data, bins=30)[1][:-1] + np.histogram(data, bins=30)[1][1:])
    
    # Plot the density curve
    plt.plot(bin_centers, density, color='red', label='Density', linewidth=2)

    # Add titles and labels
    plt.title(f'Distribution of {col}')
    plt.xlabel(f'{col} Levels')
    plt.ylabel('Density')
    plt.legend()

# Adjust layout and show the plot
plt.tight_layout()
plt.show()
```

```python
#Creating the density plots
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set the figure size
plt.figure(figsize=(12, 6))

# List of gene expression columns
columns = ['Gene_Expression', 'Gene_A_Expression', 'Gene_B_Expression']

# Loop through each column to create density plots on the same plot
for col in columns:
    sns.kdeplot(df[col], label=col, fill=True)

# Add titles and labels
plt.title('Density Plot of Gene Expression Levels')
plt.xlabel('Expression Levels')
plt.ylabel('Density')

# Add legend to differentiate between the columns
plt.legend()

# Show the plot
plt.show()

```

```python
#Creating the box plot
import matplotlib.pyplot as plt
import seaborn as sns

# Set the figure size
plt.figure(figsize=(12, 6))

# Melt the DataFrame to long format for Seaborn
df_melted = df.melt(value_vars=['Gene_Expression', 'Gene_A_Expression', 'Gene_B_Expression'],
                    var_name='Gene', value_name='Expression Level')

# Create the boxplot
sns.boxplot(x='Gene', y='Expression Level', data=df_melted, palette='Set2')

# Add titles and labels
plt.title('Boxplot of Gene Expression Levels Across Samples')
plt.xlabel('Gene')
plt.ylabel('Expression Level')

# Show the plot
plt.show()
```

```python
#Creating the violin plot
import matplotlib.pyplot as plt
import seaborn as sns

# Set the figure size
plt.figure(figsize=(12, 6))

# Melt the DataFrame to long format for Seaborn
df_melted = df.melt(value_vars=['Gene_Expression', 'Gene_A_Expression', 'Gene_B_Expression'],
                    var_name='Gene', value_name='Expression Level')

# Create the violin plot
sns.violinplot(x='Gene', y='Expression Level', data=df_melted, inner='quartile', palette='Set2')

# Add titles and labels
plt.title('Violin Plot of Gene Expression Levels Across Samples')
plt.xlabel('Gene')
plt.ylabel('Expression Level')

# Show the plot
plt.show()
```

```python
#Creating the scatter plot for two gene expressions
import matplotlib.pyplot as plt

# Generate a color mapping based on unique samples
unique_samples = df['Sample'].unique()
color_map = {sample: color for sample, color in zip(unique_samples, plt.cm.Set1.colors)}

# Map each sample to its corresponding color
colors = df['Sample'].map(color_map)

# Set the figure size for all plots
plt.figure(figsize=(12, 15))

# Scatter Plot 1: Gene_Expression vs Gene_A_Expression
plt.subplot(3, 1, 1)
scatter1 = plt.scatter(df['Gene_Expression'], df['Gene_A_Expression'], c=colors, alpha=0.7)
plt.title('Scatter Plot of Gene_Expression vs Gene_A_Expression Levels')
plt.xlabel('Gene Expression Level')
plt.ylabel('Gene A Expression Level')
plt.grid()

# Create a legend for the first plot
handles1 = [plt.Line2D([0], [0], marker='o', color=color, label=sample, linestyle='') for sample, color in color_map.items()]
plt.legend(handles=handles1, title="Samples", bbox_to_anchor=(1.05, 1), loc='upper left')

# Scatter Plot 2: Gene_Expression vs Gene_B_Expression
plt.subplot(3, 1, 2)
scatter2 = plt.scatter(df['Gene_Expression'], df['Gene_B_Expression'], c=colors, alpha=0.7)
plt.title('Scatter Plot of Gene_Expression vs Gene_B_Expression Levels')
plt.xlabel('Gene Expression Level')
plt.ylabel('Gene B Expression Level')
plt.grid()

# Create a legend for the second plot
handles2 = [plt.Line2D([0], [0], marker='o', color=color, label=sample, linestyle='') for sample, color in color_map.items()]
plt.legend(handles=handles2, title="Samples", bbox_to_anchor=(1.05, 1), loc='upper left')

# Scatter Plot 3: Gene_A_Expression vs Gene_B_Expression
plt.subplot(3, 1, 3)
scatter3 = plt.scatter(df['Gene_A_Expression'], df['Gene_B_Expression'], c=colors, alpha=0.7)
plt.title('Scatter Plot of Gene_A_Expression vs Gene_B_Expression Levels')
plt.xlabel('Gene A Expression Level')
plt.ylabel('Gene B Expression Level')
plt.grid()

# Create a legend for the third plot
handles3 = [plt.Line2D([0], [0], marker='o', color=color, label=sample, linestyle='') for sample, color in color_map.items()]
plt.legend(handles=handles3, title="Samples", bbox_to_anchor=(1.05, 1), loc='upper left')

# Adjust layout and show the plots
plt.tight_layout()
plt.show()


```
